# memory management in paython

Small, immutable objects (such as small integers or strings) are stored in the same memory location. 

```python
# small immutable integer
x = 5
```

```python
# check memory address
id(x)
```

```python
# reassign variable
x = 13
```

```python
# check memory address
id(x) # x = 13
```

```python
# check value
x
```

In the last example, it creates two different addresses

```python
# small immutable integer
y = 13 
```

```python
# check memory address
id(y)
```

In the last example, it uses the same address because int is a small immutable integer

```python
# check if booleans are the same object in memory
x is y
```

```python
# large immutable integer
a = 12548768
```

```python
# another large immutable integer
b = 12548768
```

```python
# check if a and b are the same object in memory
a is b
```

In the last example, it uses a different address because int is a large immutable integer.

```python
# mutable list
c = [3]
```

```python
# another mutable list
e = [3]
```

```python
# check if c and e are the same object in memory
c is e
```

In the last example, it uses a different address because a list is mutable.

```python
# small immutable tuple
i = (3)
```

```python
# another small immutable tuple
j = (3)
```

```python
# check if i and j are the same object in memory
i is j
```

In the last example, it uses the same address because a tuple is a small immutable object.
